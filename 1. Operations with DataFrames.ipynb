```python
# We also can access to tables by using:

denuncias_huaral_df = spark.table("workspace.default.dataset_modalidades_actuales_delimitado")
denuncias_huaral_df.show(5)
```

```python
# Using volumes is generally preferred because it provides better compatibility, clearer ​data management and more flexibility when working with files and tables in PySpark
```

```python
# display function is more useful for visualizing data in the notebook

display(denuncias_huaral_df)
```

```python
# printschema function will provide us information about the structure of our dataset

denuncias_huaral_df.printSchema()
```

```python
# We can use select function to select specific columns from our dataset

denuncias_huaral_df.select("materia_hecho", "tipo_hecho").display()
```

```python
# Note that execution only happens when an action is called (show or display)
```

```python
# We can use filter function to filter our dataset

denuncias_huaral_df.filter(denuncias_huaral_df["materia_hecho"] == "FUERO COMUN").display()
```

```python
# Aggregations also can be applied

denuncias_huaral_df.groupBy("subtipo_hecho").count().display()
```

```python
# We can use describe function to get some statistics about our dataset

denuncias_huaral_df.describe().display()

# We can use groupby function to group our dataset by a specific column

denuncias_huaral_df.groupBy("materia_hecho").count().display()

# We can use orderBy function to order our dataset by a specific column

denuncias_huaral_df.orderBy("materia_hecho").display()

# We can use withColumnRenamed function to rename a specific column

denuncias_huaral_df.withColumnRenamed("materia_hecho", "materia").display()

# We can use drop function to drop a specific column

denuncias_huaral_df.drop("materia_hecho").display()

# We can use withColumn function to add a new column to our dataset

denuncias_huaral_df.withColumn("new_column", denuncias_huaral_df["materia_hecho"] + " " + denuncias_huaral_df["tipo_hecho"]).display()

# We can use dropDuplicates function to drop duplicate rows from our dataset

denuncias_huaral_df.dropDuplicates().display()

# We can use dropna function to drop rows with null values from our dataset

denuncias_huaral_df.dropna().display()
```
