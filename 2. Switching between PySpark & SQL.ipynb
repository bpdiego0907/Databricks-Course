```python
denuncias_df = spark.read.csv("/Volumes/workspace/default/denuncias_vol/denuncias_huaral_url1.csv", header=True, inferSchema=True)

denuncias_df.display()
```

```python
denuncias_df.printSchema()
```

```python
# This command creates a temporary view (table) named "denuncias" in the current Spark session, so the data is not copied. It is storaged in memory

denuncias_df.createOrReplaceTempView("denuncias")
```

```python
%sql
SELECT modalidad_hecho, regionpol_hecho, mes_hecho
FROM denuncias
LIMIT 10
```

```python
%sql
SELECT modalidad_hecho, COUNT(*) as total_rows
FROM denuncias
GROUP BY modalidad_hecho
ORDER BY total_rows DESC
LIMIT 10
```

```python
# Now we can replicate the code in PySpark

denuncias_df.groupBy("modalidad_hecho").count().orderBy("count", ascending=False).show(10)
```

```python
# We can also use SQL

spark.sql("SELECT modalidad_hecho, count(*) as total FROM denuncias GROUP BY modalidad_hecho ORDER BY total DESC")
```
