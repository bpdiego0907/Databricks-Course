###All magic commands in Databricks initiaties with a "%" symbol like "%md", "%sql"

**%fs** 
allows us to interact with DBFS and it is commonly used to list files, check wheter a file exists, debug file path issues and if a file read operations fails it can help us to detect wheter the file path is correct

```python
# List files stored in the file store directory

%fs ls /FileStore/tables/
```

**%run** allows us to run another notebook whether our currently notebook

```python
%run "/Workspace/Users/diegoblpach09@gmail.com/Includes/Setup"
```

**%scala** allows as to write scalar code inside a notebook but some environments like Serverless or restricted cluster don't support yet this language

```python
%scala
spark.range(1,6).show()
```

**%r** allows us to execute R code but is restritecd for some environments like Scala

**%sh** allows us to run linux shell commands on the cluster

```python
%sh

ls
```

**%pip** allows us to install python packages directly inside a notebook

**%env** allows us to create environment variables which can be accesed by Python and Spark code

```python
%env
curso = 'databricks'
```

#All this magic commands communicate with the same SparkSession which lives on the driver node. That is why these different languages work seamlessly.
