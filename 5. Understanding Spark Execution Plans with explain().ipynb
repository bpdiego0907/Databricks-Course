```python
denuncias_df = spark.read.csv("/Volumes/workspace/default/denuncias_vol/denuncias_huaral_url1.csv", header=True, inferSchema=True)

denuncias_df.display()
```

```python
denuncias_filtrado = denuncias_df.filter((denuncias_df['fecha_hecho'] >= '2019-01-01') & (denuncias_df['fecha_hecho'] <= '2019-12-31'))
```

###At this point I only told it what have to do, but there is no an execution command so we can use explain() to know the plan that Spark builded for this instruction

```python
denuncias_filtrado.explain(True)
```

**Parsed logical plan** is spark understanding our code exactly as we wrote it\
**Analyzed logical plan** contains an analysis of rows and columns. Spark finds a way to do the action\
**Optimized logical plan** is where spark becomes smart. It tries to optimize your code making some transformations like removing unnecesary columns, etc. aiming to reduce processing\
**Physical plan** is the actual execution
