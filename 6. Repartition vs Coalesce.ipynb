#Repartition

```python
denuncias_df = spark.read.csv("/Volumes/workspace/default/denuncias_vol/denuncias_huaral_url1.csv", header=True, inferSchema=True)

denuncias_df.display()
```

Check number of partititons distributions

```python
# denuncias_df.rdd.getNumPartitions()

from pyspark.sql.functions import spark_partition_id

denuncias_with_pid = denuncias_df.withColumn(
    "partition_id",
    spark_partition_id()
)

display(denuncias_with_pid)
```

```python
# Number of partitions

denuncias_with_pid.select("partition_id").distinct().count()
```

```python
denuncias_with_pid.explain(True)
```

# Coalesce

```python
coalesced_df = denuncias_df.coalesce(2)
```

```python
coalesced_with_pid = coalesced_df.withColumn(
    "partition_id",
    spark_partition_id()
)
display(coalesced_with_pid)
```

```python
coalesced_with_pid.explain(True)
```

#Writing repartitioned and coalesced dataframes

```python
denuncias_with_pid.write.mode("overwrite").csv("/Volumes/workspace/default/denuncias_vol/denuncias_huaral_url1_partitioned")
```

```python
coalesced_with_pid.write.mode("overwrite").csv("/Volumes/workspace/default/denuncias_vol/denuncias_huaral_url1_coalesced")
```
