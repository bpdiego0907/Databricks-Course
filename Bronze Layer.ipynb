```python
%sql
CREATE TABLE workspace.default.walmart_dataset
USING DELTA
LOCATION 's3://databricks-bucket-packt-demo/bronze/'
AS
SELECT
    *,
    current_timestamp() AS created_at,
    input_file_name() AS source_path,
    uuid() AS record_id
FROM csv.`s3://databricks-bucket-packt-demo/Walmart_Sales.csv`
WITH (header = true, inferSchema = true);
```
